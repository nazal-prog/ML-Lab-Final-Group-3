```python
def predict_message(message):
    message_tfidf = vectorizer.transform([message])
    
    prediction = model.predict(message_tfidf)[0]
    probabilities = model.predict_proba(message_tfidf)[0]
    
    if prediction == 1:
        label = "Spam"
    else:
        label = "Ham"
    
    spam_probability = probabilities[1]
    
    return label, spam_probability
```

```python
label, spam_probability = predict_message(
    "Congratulations! You have won a free prize!"
)

print("Prediction:", label)
print("Spam probability:", spam_probability)
```

```python
label, spam_probability = predict_message(
    "Hey, are we still meeting for lunch tomorrow?"
)

print("Prediction:", label)
print("Spam probability:", spam_probability)
```

```python
test_messages = [
    "Free entry in a weekly competition! Text WIN to 80000 now!",
    "Can you send me the assignment notes?",
    "Congratulations, you have won a cash prize!",
    "I'll call you when I reach home."
]

for message in test_messages:
    label, spam_probability = predict_message(message)
    
    print("Message:", message)
    print("Prediction:", label)
    print("Spam probability:", round(spam_probability, 4))
    print("-" * 50)
```

```python
test_messages = [
    "Free entry in a weekly competition! Text WIN to 80000 now!",
    "Can you send me the assignment notes?",
    "Congratulations, you have won a cash prize!",
    "I'll call you when I reach home.",
    "URGENT! Claim your free reward now!"
]

print(test_messages)
```

```python
results = []

for message in test_messages:
    label, spam_probability = predict_message(message)
    
    results.append({
        "message": message,
        "prediction": label,
        "spam_probability": spam_probability
    })

results
```

```python
import pandas as pd

results_df = pd.DataFrame(results)

results_df
```
